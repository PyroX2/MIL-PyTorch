```python
import os
import cv2
import numpy as np
import pydicom
from ultralytics import YOLO

YOLO_MODEL_PATH = "best.pt"

def is_dicom_path(path):
    return os.path.splitext(path)[1].lower() in [".dcm", ".dicom"]

def load_image_any(path):
    if is_dicom_path(path):
        ds = pydicom.dcmread(path)
        img = ds.pixel_array
        return img, ds
    img = cv2.imread(path, cv2.IMREAD_UNCHANGED)
    return img, None

def make_working_photo(img):
    x = np.array(img)
    if x.ndim == 2:
        x3 = np.stack([x, x, x], axis=-1)
    elif x.ndim == 3 and x.shape[2] == 1:
        x3 = np.repeat(x, 3, axis=2)
    else:
        x3 = x[..., :3]

    x3 = x3.astype(np.float32)
    mn = float(x3.min())
    mx = float(x3.max())
    if mx > mn:
        x3 = (x3 - mn) / (mx - mn)
    else:
        x3[:] = 0.0
    x3 = (x3 * 255.0).clip(0, 255).astype(np.uint8)
    return x3

def get_best_box_xyxy(result):
    boxes = result.boxes
    if boxes is None or len(boxes) == 0:
        return None
    xyxy = boxes.xyxy.detach().cpu().numpy()
    if hasattr(boxes, "conf") and boxes.conf is not None:
        conf = boxes.conf.detach().cpu().numpy()
        i = int(np.argmax(conf))
    else:
        areas = (xyxy[:, 2] - xyxy[:, 0]) * (xyxy[:, 3] - xyxy[:, 1])
        i = int(np.argmax(areas))
    return xyxy[i]

def crop_on_original(img, box):
    h, w = img.shape[:2]
    x1, y1, x2, y2 = box
    x1 = int(np.floor(max(0, min(w, x1))))
    y1 = int(np.floor(max(0, min(h, y1))))
    x2 = int(np.ceil(max(0, min(w, x2))))
    y2 = int(np.ceil(max(0, min(h, y2))))
    cropped = img[y1:y2, x1:x2].copy()
    return cropped, (x1, y1, x2, y2)

def save_same_type(out_path, cropped, dicom_ds=None):
    if is_dicom_path(out_path):
        ds = dicom_ds.copy()
        ds.Rows = int(cropped.shape[0])
        ds.Columns = int(cropped.shape[1])
        ds.PixelData = cropped.tobytes()
        ds.save_as(out_path)
    else:
        cv2.imwrite(out_path, cropped)

def yolo_crop_original_image(img_path, out_path=None, model_path=YOLO_MODEL_PATH, conf=0.25, save=False):
    original_img, dicom_ds = load_image_any(img_path)
    working_img = make_working_photo(original_img)
    model = YOLO(model_path)
    result = model.predict(source=working_img, conf=conf, verbose=False)[0]
    box = get_best_box_xyxy(result)

    if box is None:
        cropped = original_img.copy()
        coords = (0, 0, original_img.shape[1], original_img.shape[0])
    else:
        cropped, coords = crop_on_original(original_img, box)

    if save:
        if out_path is None:
            base, ext = os.path.splitext(img_path)
            out_path = f"{base}_yolo_crop{ext}"
        save_same_type(out_path, cropped, dicom_ds=dicom_ds)

    return cropped, coords, out_path
```

```python
img_path = "spot_mag_dcm.dcm"
out_path = "spot_mag_dcm_crop.dcm"

cropped, coords, saved_path = yolo_crop_original_image(
    img_path=img_path,
    out_path=out_path,
    model_path="best.pt",
    conf=0.25,
    save=True
)

print("coords xyxy:", coords)
print("saved_path:", saved_path)
print("cropped shape:", cropped.shape, "dtype:", cropped.dtype)
```
