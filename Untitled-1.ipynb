Kod i algorytm do wycinania:

```python
import os
import cv2
import numpy as np
import pydicom
from pydicom.pixels import apply_voi_lut
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

#liczy najdluzsze linie prawie biale z gory i zdolu
def longest_runs_from_top(mask: np.ndarray) -> np.ndarray:
    h, w = mask.shape
    runs = np.zeros(w, dtype=np.int32)
    for x in range(w):
        col = mask[:, x]
        idx = np.where(~col)[0]
        runs[x] = (idx[0] if idx.size else h)
    return runs


def longest_runs_from_bottom(mask: np.ndarray) -> np.ndarray:
    h, w = mask.shape
    runs = np.zeros(w, dtype=np.int32)
    for x in range(w):
        col = mask[::-1, x]
        idx = np.where(~col)[0]
        runs[x] = (idx[0] if idx.size else h)
    return runs


def is_dicom_path(path: str) -> bool:
    ext = os.path.splitext(path)[1].lower()
    return ext in [".dcm", ".dicom"]


def normalize_for_analysis(img: np.ndarray, dicom_ds=None) -> np.ndarray:
    if img.ndim == 3:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32)
    else:
        gray = img.astype(np.float32).copy()

    if dicom_ds is not None:
        try:
            gray = apply_voi_lut(gray, dicom_ds).astype(np.float32)
        except Exception:
            pass

        if hasattr(dicom_ds, "BitsStored"):
            max_val = float((1 << int(dicom_ds.BitsStored)) - 1)
        else:
            max_val = float(np.max(gray))
    else:
        max_val = float(np.max(gray))

    if max_val > 0:
        gray = gray / max_val
    else:
        gray[:] = 0.0

    gray = np.clip(gray, 0.0, 1.0)
    return gray


# zmienic mozna i dotiunowac
def normalize_for_display(img: np.ndarray):
    if img.ndim == 2:
        disp = img.astype(np.float32)
        dmin = float(np.min(disp))
        dmax = float(np.max(disp))

        if dmax > dmin:
            disp = (disp - dmin) / (dmax - dmin)
        else:
            disp[:] = 0.0

        return disp

    disp = cv2.cvtColor(img, cv2.COLOR_BGR2RGB).astype(np.float32)
    dmin = float(np.min(disp))
    dmax = float(np.max(disp))

    if dmax > dmin:
        disp = (disp - dmin) / (dmax - dmin)
    else:
        disp[:] = 0.0

    return disp


def _find_row_intervals(row_mask: np.ndarray):
    intervals = []
    in_band = False
    start = 0

    for i, v in enumerate(row_mask):
        if v and not in_band:
            start = i
            in_band = True
        elif not v and in_band:
            intervals.append((start, i - 1))
            in_band = False

    if in_band:
        intervals.append((start, len(row_mask) - 1))

    return intervals


def crop_by_axis_runs(
    img,
    pre_cut_frac=0.0,
    post_cut_frac=0.0,

    # 4*0.0125= 0.05 wiec ZAWSZE usunie chociaz 5% obrazu troche sanity check by nie bylo bialych smog

    # tbh zgadywalem liczby
    zone_frac=1.0,        
    white_thresh=240,     # czym jest prawie bialy
    run_len_thresh=20,    
    min_cols_frac=0.05,   # jaki ulamek szerokosci ma byc bialy, aby uznac wiersz za pas
    use_percentile=99.5,  
    safe_margin_frac=0.025,
    min_band_height=3
):
    gray = img
    H, W = gray.shape

    # precut
    pre = int(round(pre_cut_frac * H))
    work_top = pre
    work_bottom = H - pre
    if work_bottom <= work_top + 30:
        return 0, H

    work = gray[work_top:work_bottom, :]
    h, w = work.shape

    white_thresh_scaled = white_thresh / 255.0

    # maska bardzo jasnych pikseli
    mask = work >= white_thresh_scaled

    # dla kazdego wiersza liczmy, jaki ulamek kolumn jest bialy
    row_white_frac = mask.mean(axis=1)

    # wiersz nalezy do bialego pasa, jesli duza czesc jego szerokosci jest biala
    row_mask = row_white_frac >= min_cols_frac

    # znajdz wszystkie biale poziome pasy jako pelne przedzialy wierszy
    intervals = _find_row_intervals(row_mask)

    # odfiltruj bardzo cienkie pasy
    intervals = [(a, b) for (a, b) in intervals if (b - a + 1) >= min_band_height]

    # potrzebujemy dwoch najwiekszych pasow
    if len(intervals) < 2:
        top = work_top
        bottom = work_bottom
    else:
        # wybierz DWA najwieksze pasy
        two_largest = sorted(intervals, key=lambda t: (t[1] - t[0] + 1), reverse=True)[:2]

        # z tych dwoch wybierz, ktory jest wyzej, a ktory nizej
        two_largest = sorted(two_largest, key=lambda t: t[0])
        top_band = two_largest[0]
        bottom_band = two_largest[1]

        # wycinamy obszar miedzy CALYMI pasami
        inner_top = top_band[1] + 1
        inner_bottom = bottom_band[0] - 1

        if inner_bottom <= inner_top + 30:
            top = work_top
            bottom = work_bottom
        else:
            # safe margines odsuniety od pasow do srodka
            margin = int(round(safe_margin_frac * h))
            top = work_top + min(h, inner_top + margin)
            bottom = work_top + max(0, inner_bottom - margin)

    top = int(np.clip(top, 0, H))
    bottom = int(np.clip(bottom, 0, H))
    if bottom <= top + 30:
        top, bottom = work_top, work_bottom

    # post cut
    out_h = bottom - top
    post = int(round(post_cut_frac * out_h))
    top2 = int(np.clip(top + post, 0, H))
    bottom2 = int(np.clip(bottom - post, 0, H))
    if bottom2 <= top2 + 30:
        top2, bottom2 = top, bottom

    return top2, bottom2


def load_image_any(img_path: str):
    if is_dicom_path(img_path):
        ds = pydicom.dcmread(img_path)
        img = ds.pixel_array
        return img, ds
    else:
        img = cv2.imread(img_path, cv2.IMREAD_UNCHANGED)
        return img, None


def save_same_type(out_path: str, cropped: np.ndarray, dicom_ds=None):
    ext = os.path.splitext(out_path)[1].lower()

    if ext in [".dcm", ".dicom"]:
        ds_out = dicom_ds.copy()
        ds_out.Rows, ds_out.Columns = cropped.shape[:2]
        ds_out.PixelData = cropped.tobytes()
        ds_out.save_as(out_path)
    else:
        cv2.imwrite(out_path, cropped)


def show_crop(
    img_path: str,
    *,
    save=True,
    out_path=None,
    pre_cut_frac=0.0,
    post_cut_frac=0.0,
    zone_frac=1.0,
    white_thresh=240,
    run_len_thresh=20,
    min_cols_frac=0.05,
    use_percentile=99.5,
    safe_margin_frac=0.025,
    min_band_height=3,
    figsize=(10, 6)
):
    original_img, dicom_ds = load_image_any(img_path)

    analysis_img = normalize_for_analysis(original_img, dicom_ds)
    top, bottom = crop_by_axis_runs(
        analysis_img,
        pre_cut_frac=pre_cut_frac,
        post_cut_frac=post_cut_frac,
        zone_frac=zone_frac,
        white_thresh=white_thresh,
        run_len_thresh=run_len_thresh,
        min_cols_frac=min_cols_frac,
        use_percentile=use_percentile,
        safe_margin_frac=safe_margin_frac,
        min_band_height=min_band_height
    )

    disp = normalize_for_display(original_img)

    H, W = disp.shape[:2]

    fig, ax = plt.subplots(figsize=figsize)
    ax.imshow(disp, cmap="gray" if disp.ndim == 2 else None)
    ax.add_patch(Rectangle((0, top), W, bottom - top, fill=False, linewidth=2.5, edgecolor="lime"))
    ax.set_title(f"{img_path}\nZachowany obszar: top={top}, bottom={bottom} (H={H})")
    ax.axis("off")
    plt.show()

    cropped = original_img[top:bottom, :]

    cropped_disp = normalize_for_display(cropped)

    fig, ax = plt.subplots(figsize=figsize)
    ax.imshow(cropped_disp, cmap="gray" if cropped_disp.ndim == 2 else None)
    ax.set_title("Po przycięciu (wynik)")
    ax.axis("off")
    plt.show()

    if save:
        if out_path is None:
            base, ext = os.path.splitext(img_path)
            out_path = f"{base}_cropped{ext}"
        save_same_type(out_path, cropped, dicom_ds=dicom_ds)
        print(f"Zapisano wycinek do: {out_path}")

    return cropped, (top, bottom)
```
